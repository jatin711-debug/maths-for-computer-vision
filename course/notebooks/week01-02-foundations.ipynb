```python
v = I.reshape(-1)
reconstructed = v.reshape(I.shape)
print("Row-order vector:", v)
print("Restored image:\n", reconstructed)
assert np.array_equal(I, reconstructed)
```

## Part C: Weighted combinations and composition

Two images need matching shapes and meaningful alignment. The coefficient below changes their contribution; the blending is performed on toy stored values.

```python
A = np.array([[0, 255], [0, 255]], dtype=float)
B = np.array([[0, 0], [255, 255]], dtype=float)
a = 0.5
C = a * A + (1 - a) * B
print("A:\n", A, "\nB:\n", B, "\nBlend:\n", C)
assert C.shape == A.shape == B.shape
assert np.all((C >= 0) & (C <= 255))
f = lambda x: 2 * x
g = lambda x: x + 10
print("g(f(30)):", g(f(30)), "f(g(30)):", f(g(30)))
```

## Part D: Loss, sigmoid and logarithms

A score, probability and loss are three different objects. Mean squared error is one possible regression objective. For classification, negative log likelihood relates the correct-class probability to a penalty.

```python
predictions = np.array([2.0, 4.0])
targets = np.array([1.0, 7.0])
residuals = predictions - targets
mse = np.mean(residuals ** 2)
print("Residuals:", residuals, "MSE:", mse)
assert np.isclose(mse, 5.0)
z = np.array([-np.log(3.0), 0.0, np.log(3.0)])
probabilities = 1 / (1 + np.exp(-z))  # Safe for these modest demonstration scores.
print("Scores:", z)
print("Sigmoid probabilities:", probabilities)
print("Correct-class losses:", -np.log(np.array([0.5, 0.25])))
assert np.allclose(probabilities, [0.25, 0.5, 0.75])
```
